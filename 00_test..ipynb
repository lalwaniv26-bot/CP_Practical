```python
# get marks of students and give a grade based on the marks
# 95 and above - GRADE A
# 85 and above - GRADE B
# 75 and above - GRADE C
# 65 and above - GRADE D
# 55 and above - GRADE E

marks = int(input("Enter the marks of the student: "))

if marks >= 95:
    print("Grade: A")
elif marks >= 85:
    print("Grade: B")
elif marks >= 75:
    print("Grade: C")
elif marks >= 65:
    print("Grade: D")
elif marks >= 55:
    print("Grade: E")
else:
    print("Grade: F")
```
